```python
import finnhub
import os
import time
from datetime import date, timedelta
import json
from dotenv import load_dotenv

load_dotenv()

print("===============================================")
print("   Finnhub Filings API Test Script (Verbose)   ")
print("===============================================\n")

# ---------------------------------------------------------
# 1. Set Finnhub API key
# ---------------------------------------------------------
print("[STEP 1] Loading FINNHUB_API_KEY...")

API_KEY = os.getenv("FINNHUB_API_KEY")

if not API_KEY:
    raise Exception("❌ FINNHUB_API_KEY is missing! Set it before running.")

print("✔ API key loaded.\n")

print("[STEP 2] Initializing Finnhub client...")
client = finnhub.Client(api_key=API_KEY)
print("✔ Finnhub client initialized.\n")

# ---------------------------------------------------------
# 2. Choose ticker & safe date range
# ---------------------------------------------------------
ticker = "AAPL"
lookback_days = 90

today = date.today()
from_date = today - timedelta(days=lookback_days)

print("[STEP 3] Preparing request...")
print(f"→ Ticker         : {ticker}")
print(f"→ Lookback Days  : {lookback_days}")
print(f"→ From Date      : {from_date}")
print(f"→ To Date        : {today}\n")
time.sleep(0.5)

# ---------------------------------------------------------
# 3. Fetch filings
# ---------------------------------------------------------
print("[STEP 4] Calling Finnhub API...")
print("   (This may take 1–2 seconds...)")
time.sleep(0.5)

try:
    filings = client.filings(
        symbol=ticker,
        _from=from_date.strftime("%Y-%m-%d"),
        to=today.strftime("%Y-%m-%d")
    )

    print("✔ Finnhub API call completed.\n")

    print("===============================================")
    print("                   RAW RESPONSE                ")
    print("===============================================")
    print(json.dumps(filings, indent=2))

    print("\n===============================================")
    print("                    SUMMARY                    ")
    print("===============================================")
    print(f"Total Filings Found: {len(filings)}")

    if len(filings) > 0:
        print("\nFirst Filing Object:")
        print(json.dumps(filings[0], indent=2))
        print("\n✔ SUCCESS: Filings data retrieved.")
    else:
        print("\n⚠ No filings returned in the last 90 days.")
        print("  → This may indicate:")
        print("    - API Key restrictions (free tier)")
        print("    - Rate limiting")
        print("    - Wrong dates")
        print("    - No filings in that date window")

except Exception as e:
    print("\n❌ ERROR calling Finnhub API:")
    print(e)
```
